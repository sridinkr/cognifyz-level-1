```python
#Level 2
import pandas as pd
import matplotlib.pyplot as plt
import folium
res = pd.read_csv("/Dataset .csv")
res.head()
#Task 1
print("\nDistribution of Restaurant Ratings:")
rating_counts = res['Aggregate rating'].value_counts()
print(rating_counts.head())
print("\nMost common rating:")
print(res['Aggregate rating'].mode())
print("\nAverage number of votes:")
avg_votes = res['Votes'].mean()
print(avg_votes)
#Task 2
print("\nMost common cuisine combinations:")
cuisine_combo = res['Cuisines'].value_counts()
print(cuisine_combo.head(10))
print("\nAverage rating by cuisine combination:")
combo_rating = res.groupby('Cuisines')['Aggregate rating'].mean().sort_values(ascending=False)
print(combo_rating.head(10))
#Task 3
map_restaurants = folium.Map(location=[20.59, 78.96], zoom_start=5)
map_restaurants
for index, row in res.head(100).iterrows():
    folium.Marker(
        location=[row['Latitude'], row['Longitude']],
        popup=row['Restaurant Name']
    ).add_to(map_restaurants)

map_restaurants
#Task 4
print("\nRestaurant chains in dataset:")
restaurant_chains = res['Restaurant Name'].value_counts()
print(restaurant_chains.head(10))
print("\nAverage rating of restaurant chains:")
chain_rating = res.groupby('Restaurant Name')['Aggregate rating'].mean().sort_values(ascending=False)
print(chain_rating.head(10))
print("\nMost popular restaurant chains (votes):")
chain_votes = res.groupby('Restaurant Name')['Votes'].sum().sort_values(ascending=False)
print(chain_votes.head(10))

```
